#4. Remove URLs

```python
def remove_url(text):
  pattern = re.compile(r'https?://\S+|www\.\S+')
  return pattern.sub(r'', text)
```

#5. Remove Punctuation

```python
import string
string.punctuation
```

```python
exclude = string.punctuation
```

```python
def remove_punc(text):
  for char in exclude:
    text = text.replace(char,'')
  return text
```

```python
text = "string. With. Punctuation?"
text
```

```python
import time
import string

exclude = string.punctuation
start = time.time()
print(remove_punc(text))
time1 = time.time() - start
print(time1*50000)
```

```python
#Always Uses Large Datasets
def remove_punc1(text):
  return text.translate(str.maketrans('', '', exclude))
```

```python
start = time.time()
remove_punc1(text)
time2 = time.time() - start
print(time2*50000)
time1/time2
```

```python
def chat_conversion(text):
  new_text = []
  for w in text.split():
    if w.upper() in chat_words:
      new_text.append(chat_words[w.upper()])
    else:
      new_text.append(w)
  return " ".join(new_text)
```

# 7. Removing Stopwords

```python
from nltk.corpus import stopwords
```

```python
import nltk
nltk.download('stopwords')

stopwords.words("english")
```

```python
def remove_stopwords(text):
  new_text = []

  for word in text.split():
    if word in stopwords.words('english'):
      new_text.append('')
    else:
      new_text.append(word)
  x = new_text[:]
  new_text.clear()
  return" ".join(x)
```

```python
remove_stopwords(" Probably my all-time favorite movie, a story of selflessness, sacrifice and dedication to a noble cause")
```

```python
df.head()
```

```python
df['review'].apply(remove_stopwords)
```

#Handling Emojis

```python

import re

def remove_emoji(text):
  emoji_pattern = re.compile("["
                               "\U0001F300-\U0001F5FF"
                               "\U0001F600-\U0001F64F"
                               "\U0001F680-\U0001F6FF"
                               "\U0001F900-\U0001F9FF"
                               "\U0001FA70-\U0001FAFF"
                               "\U00002600-\U000026FF"
                               "\U00002700-\U000027BF"
                               "]+",flags=re.UNICODE)
  return emoji_pattern.sub(r" ", text)

```

```python
remove_emoji(text = "Hello 😍 Python 🚀🔥 NLP 😊")
```

#Stemming

```python
from nltk.stem import PorterStemmer
```

```python
ps = PorterStemmer()
def stem_words(text):
  return" ".join([ps.stem(word) for word in text.split()])
```

```python
sample = "walk walks walking walked"
stem_words(sample)
```

```python
text = (" Probably my all-time favorite movie, a story of selflessness, sacrifice and dedication to a noble cause")
text
```

#Lemmatizer

```python
import nltk
from nltk.stem import WordNetLemmatizer
wordnet_lemmatizer = WordNetLemmatizer()
```

```python
sentence = "He was running and eating at same time"
punctuation = "?:.,;"
```

```python
sentence_words = nltk.word_tokenize(sentence)
for word in sentence_words:
  if word in punctuation:
    sentence_words.remove(word)
sentence_words
```

```python
import nltk
nltk.download('wordnet')

print("{0:20}{1:20}".format("word","lemma"))
for word in sentence_words:
  print("{0:20}{1:20}".format(word,wordnet_lemmatizer.lemmatize(word)))
```
