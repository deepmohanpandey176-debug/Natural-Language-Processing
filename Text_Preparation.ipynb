#Removing HTML Tags

```python
sample_text = '<h2>HTML Element <h2><p>The HTML <code class="w3-codespan">&lt;sup&gt;,</code> elements Define super script text. Superscript text appears half a character above the normal line, like WWW/<sup>[1]</sup>:</p> '
```

```python
sample_text
```

```python
import re
def striphtml(data):
    p = re.compile(r'<.*?>')
    return p.sub(" ", data)
```

```python
striphtml(sample_text)
```
